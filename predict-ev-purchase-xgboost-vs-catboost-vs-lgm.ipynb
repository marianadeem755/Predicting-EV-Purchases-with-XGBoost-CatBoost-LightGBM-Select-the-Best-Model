# <p style="width:750px; border-radius:20px; border:8px solid #9A6415; padding:22px 28px; background:linear-gradient(135deg,#FFF4D6,#FFD98A,#B7791F); text-align:center; box-shadow:0 0 0 3px #E8B84A, 0 10px 25px rgba(90,55,5,0.32), inset 0 2px 0 rgba(255,255,255,0.75); color:#3B2608; text-shadow:0 1px 1px rgba(255,255,255,0.4); font-weight:800; margin:28px auto; font-size:25px; letter-spacing:0.5px;">Lets Do Advanced Feature Engineering</p>

```python
# Feature Engineering Function
def fit_feature_engineering(X_train):
    category_map, bin_config = {}, {"Annual_Income_USD": [400, 600, 800, 900, 1100]}
    for col, bins_list in bin_config.items():
        for n_bins in bins_list:
            bin_name = f"{col}_bin_{n_bins}"
            kb = KBinsDiscretizer(n_bins=n_bins, encode="ordinal", strategy="quantile", subsample=None)
            kb.fit(X_train[[col]])
            category_map[bin_name] = kb
    return category_map
# Apply Feature Engineering
def apply_feature_engineering(df, num_features_, category_map):
    df = df.copy()
    num_to_cat = []
    for col in num_features_:
        cat_name = f"{col}_cat"
        df[cat_name] = df[col].fillna("NaN").astype(str)
        num_to_cat.append(cat_name)
    bin_config = {"Annual_Income_USD": [400, 600, 800, 900, 1100]}
    for col, bins_list in bin_config.items():
        for n_bins in bins_list:
            bin_name = f"{col}_bin_{n_bins}"
            df[bin_name] = pd.Series(category_map[bin_name].transform(df[[col]]).ravel().astype("int32"),index=df.index).astype("category")
    for c in num_features_:
        for k in range(-4, 4):
            df[f"{c}_digit{k}"] = (df[c].fillna(0) // (10 ** k) % 10).astype("int8")
    df["index"] = df["Environmental_Concern_Level"] * df["Charging_Stations_Near_Home"]
    df["is_30k_spike"] = (df["Annual_Income_USD"] == 30000.0).astype("int8")
    df["is_millionaire_cliff"] = (df["Annual_Income_USD"] >= 169972).astype("int8")
    df["is_dead_zone"] = ((df["Annual_Income_USD"] >= 31003) & (df["Annual_Income_USD"] <= 41970)).astype("int8")
    df["is_env_hater"] = (df["Environmental_Concern_Level"] == 1).astype("int8")
    df["income100_floor"] = np.floor(df["Annual_Income_USD"] / 100.0).astype(str)
    df["income1000_floor"] = np.floor(df["Annual_Income_USD"] / 1000.0).astype(str)
    df["commute_integer"] = np.floor(df["Daily_Commute_km"]).astype(str)
    return df, num_to_cat
```

# <p style="width:750px; border-radius:20px; border:8px solid #9A6415; padding:22px 28px; background:linear-gradient(135deg,#FFF4D6,#FFD98A,#B7791F); text-align:center; box-shadow:0 0 0 3px #E8B84A, 0 10px 25px rgba(90,55,5,0.32), inset 0 2px 0 rgba(255,255,255,0.75); color:#3B2608; text-shadow:0 1px 1px rgba(255,255,255,0.4); font-weight:800; margin:28px auto; font-size:25px; letter-spacing:0.5px;">Lets Do Feature Encoding</p>

```python
# Frequency Encoding Functions
def fit_frequency_encoding(X, cat_features_, num_to_cat):
    all_cats = list(dict.fromkeys(cat_features_ + num_to_cat))
    fe_cols = [c for c in all_cats if c in X.columns]
    freq_encodings = {c: X[c].value_counts(normalize=True).to_dict() for c in fe_cols}
    return fe_cols, freq_encodings

def apply_frequency_encoding(X, fe_cols, freq_encodings):
    X = X.copy()
    for c in fe_cols:
        X[f"{c}_fe"] = X[c].map(freq_encodings.get(c, {})).astype(float).fillna(0.0) if c in X.columns else 0.0
    return X
```

# <p style="width:750px; border-radius:20px; border:8px solid #9A6415; padding:22px 28px; background:linear-gradient(135deg,#FFF4D6,#FFD98A,#B7791F); text-align:center; box-shadow:0 0 0 3px #E8B84A, 0 10px 25px rgba(90,55,5,0.32), inset 0 2px 0 rgba(255,255,255,0.75); color:#3B2608; text-shadow:0 1px 1px rgba(255,255,255,0.4); font-weight:800; margin:28px auto; font-size:25px; letter-spacing:0.5px;">Lets Do Feature Encoding</p>

```python
# Apply Feature Engineering & Encoding
num_features_ = X.select_dtypes(exclude=["object", "bool", "category"]).columns.tolist()
cat_features_ = X.select_dtypes(include=["object", "bool", "category"]).columns.tolist()
category_map = fit_feature_engineering(X)

X_fe, num_to_cat = apply_feature_engineering(X, num_features_, category_map)

fe_cols, freq_encodings = fit_frequency_encoding(X_fe, cat_features_, num_to_cat)
X_fe = apply_frequency_encoding(X_fe, fe_cols, freq_encodings)

# Re-select cat/num features
num_features = X_fe.select_dtypes(exclude=["object", "bool", "category"]).columns.tolist()
cat_features = X_fe.select_dtypes(include=["object", "bool", "category"]).columns.tolist()

# Ensure categorical features have consistent missing-value handling
for c in cat_features:
    X_fe[c] = X_fe[c].astype(object).fillna("NaN").astype("category")

# Same for testing data
test_fe, _ = apply_feature_engineering(test, num_features_, category_map)
test_fe = apply_frequency_encoding(test_fe, fe_cols, freq_encodings)

for c in cat_features:
    if c in test_fe.columns:
        test_fe[c] = test_fe[c].astype(object).fillna("NaN").astype("category")

print("After FE Training Data:", X_fe.shape, "& Test:", test_fe.shape)
print("Cat features:", len(cat_features), "& Num features:", len(num_features))
```

# <p style="width:750px; border-radius:20px; border:8px solid #9A6415; padding:22px 28px; background:linear-gradient(135deg,#FFF4D6,#FFD98A,#B7791F); text-align:center; box-shadow:0 0 0 3px #E8B84A, 0 10px 25px rgba(90,55,5,0.32), inset 0 2px 0 rgba(255,255,255,0.75); color:#3B2608; text-shadow:0 1px 1px rgba(255,255,255,0.4); font-weight:800; margin:28px auto; font-size:25px; letter-spacing:0.5px;">Lets Do Model Training & Cross-Validation</p>

```python
# Train XGB, LGBM & CatBoost 
folds = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=STATE)
oofs, tests, fold_scores = {}, {}, {}
models_list = [("XGB", XGBClassifier, xgb_params), ("LGBM", LGBMClassifier, lgb_params), ("CatBoost", CatBoostClassifier, cb_params)]

for name, ModelClass, params in models_list:
    print(f"\n{'='*60}\nTraining {name} ...\n{'='*60}")
    oof = np.zeros(X_fe.shape[0], dtype=np.float32)
    test_p = np.zeros(test_fe.shape[0], dtype=np.float32)
    fold_scores[name] = []
    
    for fold_idx, (train_idx, valid_idx) in enumerate(folds.split(X_fe, y)):
        print(f"\n  → {name} | Fold {fold_idx+1}/{N_SPLITS}")
        X_train, y_train = X_fe.iloc[train_idx].copy(), y.iloc[train_idx]
        X_val, y_val = X_fe.iloc[valid_idx].copy(), y.iloc[valid_idx]
        X_test = test_fe.copy()
        
        te_cols = cat_features + [c for c in X_train.columns if "_digit" in c]
        for sm in [10, 100, "auto"]:
            te = TargetEncoder(smooth=sm, random_state=42)
            X_train_enc = te.fit_transform(X_train[te_cols].reset_index(drop=True), y_train).astype("float32")
            X_val_enc   = te.transform(X_val[te_cols].reset_index(drop=True)).astype("float32")
            X_test_enc  = te.transform(X_test[te_cols].reset_index(drop=True)).astype("float32")
            te_names = [f"{col}_TE_s{sm}" for col in te_cols]
            X_train[te_names], X_val[te_names], X_test[te_names] = X_train_enc, X_val_enc, X_test_enc
        
        X_train = X_train.drop(columns=cat_features)
        X_val   = X_val.drop(columns=cat_features)
        X_test  = X_test.drop(columns=cat_features)
        
        model = ModelClass(**params)
        if name == "XGB":
            model.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)
        elif name == "LGBM":
            model.fit(X_train, y_train, eval_set=[(X_val, y_val)], eval_metric="auc", callbacks=[])
        else:
            model.fit(X_train, y_train, eval_set=(X_val, y_val), early_stopping_rounds=EARLY_STOP, verbose=0)
        
        val_pred = model.predict_proba(X_val)[:, 1]
        oof[valid_idx] = val_pred
        test_p += model.predict_proba(X_test)[:, 1] / N_SPLITS
        
        fold_auc = roc_auc_score(y_val, val_pred)
        fold_scores[name].append(fold_auc)
        print(f" {name} Fold {fold_idx+1} AUC: {fold_auc:.7f}")
        
        del X_train, X_val, X_test, model, te
        gc.collect()
    
    oofs[name], tests[name] = oof, test_p
    mean_oof = roc_auc_score(y, oof)
    print(f"\n{name} Mean OOF AUC: {mean_oof:.7f}")
    print(f"{name} Fold AUCs : {[round(s, 7) for s in fold_scores[name]]}")
```

# <p style="width:750px; border-radius:20px; border:8px solid #9A6415; padding:22px 28px; background:linear-gradient(135deg,#FFF4D6,#FFD98A,#B7791F); text-align:center; box-shadow:0 0 0 3px #E8B84A, 0 10px 25px rgba(90,55,5,0.32), inset 0 2px 0 rgba(255,255,255,0.75); color:#3B2608; text-shadow:0 1px 1px rgba(255,255,255,0.4); font-weight:800; margin:28px auto; font-size:25px; letter-spacing:0.5px;">Lets Display Model Results & Select Best Model</p>

```python
# Results summary & select best model
print("=======================================================================")
print("DETAILED FOLD RESULTS")
print("=======================================================================")

for name in fold_scores:
    print(f"\n{name}:")
    for i, score in enumerate(fold_scores[name], 1):
        print(f"  Fold {i}: {score:.7f}")
    print(f"  Mean OOF: {roc_auc_score(y, oofs[name]):.7f}")

print("\n" + "=" * 70)
print("MODEL RESULTS (Overall OOF)")
print("======================================================================")

model_scores = {
    name: roc_auc_score(y, pred)
    for name, pred in oofs.items()
}
for name, score in model_scores.items():
    print(f"{name:<12} OOF AUC: {score:.7f}")

best_name = max(model_scores, key=model_scores.get)
best_test_pred = tests[best_name]
best_score = model_scores[best_name]

print("=======================================================================")
print(f"BEST MODEL: {best_name}  (OOF AUC: {best_score:.7f})")
print("==========================================================================")
```
