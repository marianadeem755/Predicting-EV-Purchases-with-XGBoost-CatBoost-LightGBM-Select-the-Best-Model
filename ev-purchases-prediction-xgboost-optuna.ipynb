```python
# Feature Engineering
def build_features(df, fitted_bins=None, fit_mode=False):
    out = df.copy()
    # Identify numerical columns
    num_cols = out.select_dtypes(exclude=["object", "bool", "category"]).columns.tolist()
    for col in num_cols:
        out[f"{col}_str"] = out[col].fillna("missing").astype(str)
    bin_counts = [400, 600, 800, 900, 1100]
    if fit_mode:
        fitted_bins = {}
        for n_bins in bin_counts:
            binner = KBinsDiscretizer(n_bins=n_bins, encode="ordinal", strategy="quantile", subsample=None)
            binner.fit(out[["Annual_Income_USD"]])
            fitted_bins[n_bins] = binner
    for n_bins in bin_counts:
        col_name = f"Annual_Income_USD_bin_{n_bins}"
        out[col_name] = fitted_bins[n_bins].transform(out[["Annual_Income_USD"]]).ravel().astype("int32")
        out[col_name] = out[col_name].astype("category")
    for col in num_cols:
        for power in range(-4, 4):
            out[f"{col}_digit{power}"] = (out[col].fillna(0) // (10 ** power) % 10).astype("int8")
    out["env_x_home"] = out["Environmental_Concern_Level"] * out["Charging_Stations_Near_Home"]
    out["is_30k"] = (out["Annual_Income_USD"] == 30000.0).astype("int8")
    out["is_high_income"] = (out["Annual_Income_USD"] >= 169972).astype("int8")
    out["is_deadzone"] = ((out["Annual_Income_USD"] >= 31003) & (out["Annual_Income_USD"] <= 41970)).astype("int8")
    out["is_low_env"] = (out["Environmental_Concern_Level"] == 1).astype("int8")
    out["income_100"] = np.floor(out["Annual_Income_USD"] / 100).astype(str)
    out["income_1000"] = np.floor(out["Annual_Income_USD"] / 1000).astype(str)
    out["commute_floor"] = np.floor(out["Daily_Commute_km"]).astype(str)
    out["stations_total"] = out["Charging_Stations_Near_Home"] + out["Charging_Stations_Near_Work"]
    out["stations_ratio"] = out["Charging_Stations_Near_Home"] / (out["Charging_Stations_Near_Work"] + 1)
    out["income_per_car"] = out["Annual_Income_USD"] / (out["Number_of_Cars_Owned"] + 1)
    out["commute_x_env"] = out["Daily_Commute_km"] * out["Environmental_Concern_Level"]
    return out, fitted_bins

# Frequency Encoding
def add_freq_encoding(df, freq_maps=None, fit_mode=False, cat_list=None):
    out = df.copy()
    if fit_mode:
        freq_maps = {c: out[c].value_counts(normalize=True).to_dict() for c in cat_list if c in out.columns}
    for c, mp in freq_maps.items():
        out[f"{c}_freq"] = out[c].map(mp).astype(float).fillna(0.0) if c in out.columns else 0.0
    return out, freq_maps

# External Mean Encoding
def add_ext_means(df, external, target, fit_mode=False, stats=None):
    out = df.copy()
    if fit_mode:
        stats = {"__global__": float(external[target].mean())}
        for c in external.columns:
            if c != target:
                stats[c] = external.groupby(c, observed=False)[target].mean()
    gmean = stats["__global__"]
    for c, s in stats.items():
        if c == "__global__": continue
        out[f"{c}_ext_mean"] = out[c].map(s).astype(float).fillna(gmean) if (c in out.columns and not s.empty) else gmean
    return out, stats
```

# <p style="width:100%;box-sizing:border-box;border-radius:30px 0 30px 0;border:3px solid #8F2945;padding:3px;background:linear-gradient(135deg,#7A1F35,#A83D58,#7A1F35);margin:28px auto;box-shadow:0 4px 14px rgba(122,31,53,0.3);"><span style="display:block;background:linear-gradient(145deg,#10152D,#15152F);border-radius:26px 0 26px 0;padding:24px 20px;text-align:center;color:#FFFFFF;font-family:Verdana,sans-serif;font-size:25px;font-weight:800;line-height:1.5;">Preparing Features for Model Training</span></p>

```python
labels = train_df[label_col]
features = train_df.drop(columns=[label_col])
test_features = test_df.copy()
# Build engineered features
features, bins = build_features(features, fit_mode=True)
test_features, _ = build_features(test_features, fitted_bins=bins, fit_mode=False)
features, ext_stats = add_ext_means(features, external_df, label_col, fit_mode=True)
test_features, _ = add_ext_means(test_features, external_df, label_col, fit_mode=False, stats=ext_stats)
# Identify categorical columns and apply frequency encoding
cat_cols = features.select_dtypes(include=["object", "bool", "category"]).columns.tolist()
features, freq_maps = add_freq_encoding(features, fit_mode=True, cat_list=cat_cols)
test_features, _ = add_freq_encoding(test_features, freq_maps=freq_maps, fit_mode=False)

final_cat = features.select_dtypes(include=["object", "bool", "category"]).columns.tolist()
for c in final_cat:
    features[c] = features[c].astype(object).fillna("NaN").astype("category")
    test_features[c] = test_features[c].astype(object).fillna("NaN").astype("category")

print("Features Shape:", features.shape)
```

# <p style="width:100%;box-sizing:border-box;border-radius:30px 0 30px 0;border:3px solid #8F2945;padding:3px;background:linear-gradient(135deg,#7A1F35,#A83D58,#7A1F35);margin:28px auto;box-shadow:0 4px 14px rgba(122,31,53,0.3);"><span style="display:block;background:linear-gradient(145deg,#10152D,#15152F);border-radius:26px 0 26px 0;padding:24px 20px;text-align:center;color:#FFFFFF;font-family:Verdana,sans-serif;font-size:25px;font-weight:800;line-height:1.5;">XGBoost Hyperparameter Optimization with Optuna & Cross-Validation</span></p>

```python
def objective(trial):
    params = {
        # Define the XGBoost hyperparameters to be optimized
        "n_estimators": 8000,
        "learning_rate": trial.suggest_float("learning_rate", 0.006, 0.014),
        "max_depth": trial.suggest_int("max_depth", 4, 6),
        "min_child_weight": trial.suggest_int("min_child_weight", 1, 5),
        "subsample": trial.suggest_float("subsample", 0.85, 0.97),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.65, 0.85),
        "reg_lambda": trial.suggest_float("reg_lambda", 3.0, 12.0),
        "reg_alpha": trial.suggest_float("reg_alpha", 0.5, 4.0),
        "max_bin": trial.suggest_int("max_bin", 2500, 5500),
        "random_state": seed_value,
        "objective": "binary:logistic",
        "eval_metric": "auc",
        "enable_categorical": True,
        "device": "cuda",
        "early_stopping_rounds": 250,
    }
    print(f"\n{'='*55}")
    print(f"Trial {trial.number}")
    print(f"{'='*55}")
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed_value)
    oof = np.zeros(len(features))
    fold_scores = []
    te_cols = final_cat + [c for c in features.columns if "_digit" in c]
    # Perform cross-validation
    for fold, (tr_idx, va_idx) in enumerate(cv.split(features, labels)):
        X_tr = features.iloc[tr_idx].copy()
        y_tr = labels.iloc[tr_idx]
        X_va = features.iloc[va_idx].copy()
        y_va = labels.iloc[va_idx]
        for sm in [10, 100, "auto"]:
            te = TargetEncoder(smooth=sm, random_state=seed_value)
            X_tr[[f"{c}_te{sm}" for c in te_cols]] = te.fit_transform(X_tr[te_cols], y_tr).astype("float32")
            X_va[[f"{c}_te{sm}" for c in te_cols]] = te.transform(X_va[te_cols]).astype("float32")
        X_tr = X_tr.drop(columns=final_cat)
        X_va = X_va.drop(columns=final_cat)
        model = XGBClassifier(**params)
        model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], verbose=False)
        pred = model.predict_proba(X_va)[:, 1]
        oof[va_idx] = pred
        fold_auc = roc_auc_score(y_va, pred)
        fold_scores.append(fold_auc)
        print(f"  Fold {fold+1}/5 → {fold_auc:.7f}")
        del X_tr, X_va, model
        gc.collect()

    mean_auc = np.mean(fold_scores)
    print(f"Trial {trial.number} Mean AUC: {mean_auc:.7f}")
    trial.set_user_attr("fold_scores", fold_scores)
    return mean_auc
```

# <p style="width:100%;box-sizing:border-box;border-radius:30px 0 30px 0;border:3px solid #8F2945;padding:3px;background:linear-gradient(135deg,#7A1F35,#A83D58,#7A1F35);margin:28px auto;box-shadow:0 4px 14px rgba(122,31,53,0.3);"><span style="display:block;background:linear-gradient(145deg,#10152D,#15152F);border-radius:26px 0 26px 0;padding:24px 20px;text-align:center;color:#FFFFFF;font-family:Verdana,sans-serif;font-size:25px;font-weight:800;line-height:1.5;">Optuna Best Hyperparameters</span></p>

```python
# Create an Optuna study to maximize the cross-validation AUC
study = optuna.create_study(direction="maximize", sampler=TPESampler(seed=seed_value))
study.optimize(objective, n_trials=8, show_progress_bar=True)
print("================================================================")
print("Best Trial :", study.best_trial.number)
print("Best CV AUC:", round(study.best_value, 7))
```

# <p style="width:100%;box-sizing:border-box;border-radius:30px 0 30px 0;border:3px solid #8F2945;padding:3px;background:linear-gradient(135deg,#7A1F35,#A83D58,#7A1F35);margin:28px auto;box-shadow:0 4px 14px rgba(122,31,53,0.3);"><span style="display:block;background:linear-gradient(145deg,#10152D,#15152F);border-radius:26px 0 26px 0;padding:24px 20px;text-align:center;color:#FFFFFF;font-family:Verdana,sans-serif;font-size:25px;font-weight:800;line-height:1.5;">Final Parameters</span></p>

```python
best_params = {
    "n_estimators": 60000,
    "learning_rate": study.best_params["learning_rate"],
    "max_depth": study.best_params["max_depth"],
    "min_child_weight": study.best_params["min_child_weight"],
    "subsample": study.best_params["subsample"],
    "colsample_bytree": study.best_params["colsample_bytree"],
    "reg_lambda": study.best_params["reg_lambda"],
    "reg_alpha": study.best_params["reg_alpha"],
    "max_bin": study.best_params["max_bin"],
    "objective": "binary:logistic",
    "eval_metric": "auc",
    "enable_categorical": True,
    "device": "cuda",
    "early_stopping_rounds": 400,
}
```

# <p style="width:100%;box-sizing:border-box;border-radius:30px 0 30px 0;border:3px solid #8F2945;padding:3px;background:linear-gradient(135deg,#7A1F35,#A83D58,#7A1F35);margin:28px auto;box-shadow:0 4px 14px rgba(122,31,53,0.3);"><span style="display:block;background:linear-gradient(145deg,#10152D,#15152F);border-radius:26px 0 26px 0;padding:24px 20px;text-align:center;color:#FFFFFF;font-family:Verdana,sans-serif;font-size:25px;font-weight:800;line-height:1.5;">XGBoost Model Training & Cross-Validation</span></p>

```python
# Train one XGBoost model using random seed and stratified cross-validation
def train_one_seed(params, seed, features, labels, test_features, final_cat):
    print(f"\n{'='*50}")
    print(f"Training with seed = {seed}")
    print(f"{'='*50}")
    params = params.copy()
    params["random_state"] = seed
    cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=seed)
    oof = np.zeros(len(features))
    test_pred = np.zeros(len(test_features))
    scores = []
    te_cols = final_cat + [c for c in features.columns if "_digit" in c]
    for fold, (tr_idx, va_idx) in enumerate(cv.split(features, labels)):
        X_tr = features.iloc[tr_idx].copy()
        y_tr = labels.iloc[tr_idx]
        X_va = features.iloc[va_idx].copy()
        y_va = labels.iloc[va_idx]
        X_te = test_features.copy()
        for sm in [10, 100, "auto"]:
            te = TargetEncoder(smooth=sm, random_state=seed)
            X_tr[[f"{c}_te{sm}" for c in te_cols]] = te.fit_transform(X_tr[te_cols], y_tr).astype("float32")
            X_va[[f"{c}_te{sm}" for c in te_cols]] = te.transform(X_va[te_cols]).astype("float32")
            X_te[[f"{c}_te{sm}" for c in te_cols]] = te.transform(X_te[te_cols]).astype("float32")
        X_tr = X_tr.drop(columns=final_cat)
        X_va = X_va.drop(columns=final_cat)
        X_te = X_te.drop(columns=final_cat)
        model = XGBClassifier(**params)
        model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], verbose=False)
        oof[va_idx] = model.predict_proba(X_va)[:, 1]
        test_pred += model.predict_proba(X_te)[:, 1] / 10
        fold_auc = roc_auc_score(y_va, oof[va_idx])
        scores.append(fold_auc)
        # Display the overall cross-validation result 
        print(f"  Seed {seed} | Fold {fold+1}/10 → {fold_auc:.6f}")
        del X_tr, X_va, X_te, model
        gc.collect()

    mean_auc = np.mean(scores)
    print(f"\nSeed {seed} Mean CV AUC: {mean_auc:.7f}")
    return oof, test_pred, mean_auc
```

```python
seeds = [42, 2024]  
all_test_preds = []
all_oof_preds = []
all_scores = []
# Train the model using seeds
for sd in seeds:
    oof, test_pred, score = train_one_seed(
        best_params, sd, features, labels, test_features, final_cat
    )
    all_oof_preds.append(oof)
    all_test_preds.append(test_pred)
    all_scores.append(score)

final_oof = np.mean(all_oof_preds, axis=0)
final_test = np.mean(all_test_preds, axis=0)

print("===================================================================")
print("SEED AVERAGING COMPLETED")
print("="*60)
print(f"Individual Seed Scores : {[round(s, 7) for s in all_scores]}")
print(f"Average CV AUC : {np.mean(all_scores):.7f}")
# Calculate and display the final OOF AUC
print(f"Final OOF AUC  : {roc_auc_score(labels, final_oof):.7f}")
```
